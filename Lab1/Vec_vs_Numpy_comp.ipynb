```python
import numpy as np   # NumPy aims to provide an array object that is up to 50x faster than traditional Python lists.
import time

sizes = [2000, 4000, 6000, 8000, 10000, 12000, 14000, 16000]

for n in sizes:

    v1 = np.arange(n)   # creates an array of evenly spaced values within a given interval(basically builtin range() function
    v2 = np.arange(n)

    start = time.perf_counter()
    v3 = v1 + v2
    end = time.perf_counter()
    print("NumPy Addition of", n, ":", end - start)
    print(" ")

    start = time.perf_counter()
    v4 = v1 * 5
    end = time.perf_counter()
    print("NumPy Scalar Multiplication of", n, ":", end - start)
    print(" ")

    start = time.perf_counter()
    norm_value = np.linalg.norm(v2)
    end = time.perf_counter()
    print("NumPy Norm of", n, ":", end - start)
    print(" ")

    start = time.perf_counter()
    v5 = v1 - v2
    end = time.perf_counter()
    print("NumPy Subtraction of", n, ":", end - start)
    print(" ")

    start = time.perf_counter()
    v5 = -v5
    end = time.perf_counter()
    print("NumPy Negation of", n, ":", end - start)
    print(" ")

    start = time.perf_counter()
    v6 = np.ones(n)
    end = time.perf_counter()
    print("NumPy Ones of", n, ":", end - start)
    print(" ")

    start = time.perf_counter()
    v7 = np.zeros(n)
    end = time.perf_counter()
    print("NumPy Zeros of", n, ":", end - start)
    print(" ")
    print("---------------------------------------------------------------")
    print(" ")
```

#The performance results clearly show that NumPy performs significantly better than the user-defined Vec implementation for almost all vector operations.
